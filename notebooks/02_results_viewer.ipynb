# 02 - Results viewer

Run this while experiments are running or after they finish. Change `RESULTS` to `"results_quick"` (or a tagged folder such as `"results_quick_v2"`) to look at the quick check.

```python
import sys
from pathlib import Path
ROOT = Path.cwd() if (Path.cwd() / "config.py").exists() else Path.cwd().parent
sys.path.insert(0, str(ROOT))
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import config
from hfwnpo_aml import datasets
pd.set_option("display.width", 160)
print("project:", ROOT)
from IPython.display import Image, display
RESULTS = ROOT / "results"          # or ROOT / "results_quick"
FIGURES = ROOT / RESULTS.name.replace("results", "figures", 1)   # results_quick_v2 -> figures_quick_v2
```

## Live progress (last 30 lines of the log)

```python
log = RESULTS / "progress.log"
print("".join(log.read_text(encoding="utf-8").splitlines(True)[-30:]) if log.exists() else "no log yet")
```

## Summary tables (mean over CV folds)

```python
for task in ["diagnosis", "prognosis", "chemoresistance/Cytarabine", "chemoresistance/Venetoclax"]:
    f = RESULTS / task / "summary.csv"
    if f.exists():
        s = pd.read_csv(f, index_col=0)
        print(f"\n=== {task} ===")
        display(s[[c for c in s.columns if c.endswith("_mean") or c == "n_genes_mean"]])
```

## External validation and statistics

```python
for task in ["diagnosis", "prognosis"]:
    for name in ["external_validation.csv", "statistical_tests.csv"]:
        f = RESULTS / task / name
        if f.exists():
            print(f"\n=== {task}: {name} ===")
            display(pd.read_csv(f).drop(columns=["genes"], errors="ignore").round(4))
```

## Figures

```python
for png in sorted(FIGURES.glob("*.png")):
    print(png.name)
    display(Image(filename=str(png), width=520))
```
